```python
opened_red_data = open('/content/winequality-red.csv','r')
opened_white_data = open('/content/winequality-white.csv','r')
red_data_lines = opened_red_data.readlines()
opened_red_data.close()
white_data_lines = opened_white_data.readlines()
opened_white_data.close()
red_data_lines.pop(0)
white_data_lines.pop(0)

all_data_lines = red_data_lines + white_data_lines
```

```python
for i in range(len(all_data_lines)):
  all_data_lines[i] = all_data_lines[i].split(';')

fixed_acidity = []

for row in all_data_lines:
  fixed_acidity.append(float(row[0]))

print(fixed_acidity)
```

```python
fixed_acidity.sort()
```

```python
print(len(fixed_acidity))
```

```python
print(sum(fixed_acidity))
```

```python
fixed_acidity_mean = sum(fixed_acidity) / len(fixed_acidity)
# calculate the mean
print('the mean is:', fixed_acidity_mean)
```

```python
# the middle index is 3248 (the 3249th number)

fixed_acidity_median = fixed_acidity[3248]
print('the median is:',fixed_acidity_median)
```

```python
fixed_acidity_min = min(fixed_acidity)
print('the minimum is:', fixed_acidity_min)
fixed_acidity_max = max(fixed_acidity)
print('the maximum is:', fixed_acidity_max)
# calculate the min and max
```

Bonus Task

```python
# minus min then divide by range
for i in range(len(fixed_acidity)):
  fixed_acidity[i] = (fixed_acidity[i] - fixed_acidity_min)/(fixed_acidity_max - fixed_acidity_min)
print('newly scaled fixed acidity is: ',fixed_acidity)
```
