```python
# Load the wine data from saved directory
opened_red_data = open('/content/winequality-red.csv','r')
opened_white_data = open('/content/winequality-white.csv','r')
```

```python
# Read data from the loaded files
red_data_lines = opened_red_data.readlines()
opened_red_data.close()
white_data_lines = opened_white_data.readlines()
opened_white_data.close()
```

```python
# Lets take a look at one variable from our dataset and see how does the data look like
red_data_lines.pop(0)
white_data_lines.pop(0)
all_data_lines = red_data_lines + white_data_lines
for i in range(len(all_data_lines)):
  all_data_lines[i] = all_data_lines[i].split(';')
fixed_acidity = []
for row in all_data_lines:
  fixed_acidity.append(float(row[0])) # we successfully loaded the selected feature inside "fixed_acidity"
```
