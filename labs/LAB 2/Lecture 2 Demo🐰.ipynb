```python
names_list = []
namer = ""

while namer != "stop":
  namer = input("Enter your name: ")
  names_list.append(namer)

print(names_list[:-1])

file = open("namers.txt", "w")
for namer in names_list:
  if namer != "stop":
    file.write(namer + "\n")
file.close()
```

```python
file = open("namers.txt", "r")
print(file.readlines())
file.close()
```
