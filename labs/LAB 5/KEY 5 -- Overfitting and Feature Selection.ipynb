```python
# standard reading procedure
opened_iris = open("/content/modified_iris.csv",'r')
iris_lines = opened_iris.readlines()
opened_iris.close()

# lets look at the first 5 lines
iris_lines[:5]
```

```python
# this removes first row of names
iris_lines.pop(0)
print(iris_lines)
```

```python
# Once again, this is more or less standard operating procedure; just a bit condensed
iris_data = []
for line in iris_lines:
  line = line[:-1]
  line = line.split(',')
  line.pop(0)
  processed_line = [ float(i) for i in line ]
  processed_line[-1] = int(processed_line[-1])
  processed_line[-2] = int(processed_line[-2])
  iris_data.append(processed_line)
```

```python
import pandas as pd

df = pd.DataFrame(iris_data)
df
```

```python
df.columns = ["petal length", "petal width", "time when measured", "cur weather", "species"]
df.head()
```

```python
# let's shuffle it (run once)
df = df.sample(frac=1, random_state=50)
df = df.reset_index(drop=True)
df.head()
```

```python
# split into train and test
X = df.iloc[:,:-1]
y = df.iloc[:,-1]
train_X = X.iloc[0:100]
train_y = y.iloc[0:100]
test_X = X.iloc[100:150]
test_y = y.iloc[100:150]
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
# fit log reg model
log_reg = LogisticRegression(max_iter = 1000)
log_reg = log_reg.fit(train_X,train_y)
```

```python
log_reg.score(test_X,test_y)
```

```python
print("Model predictions:", list(log_reg.predict(test_X)))
print("Ground truth:     ", list(test_y))
```

```python
from sklearn.feature_selection import SelectKBest, chi2
filter = SelectKBest(chi2, k=2)
# we just want to fit on training data
filter.fit(train_X, train_y)
# but apply the same transofrmation to both datasets
new_train_X = filter.transform(train_X)
new_test_X = filter.transform(test_X)
```

```python
# convert everything to dataframes
new_train_X = pd.DataFrame(new_train_X)
new_test_X = pd.DataFrame(new_test_X)
```

```python
remaining_features = filter.get_feature_names_out()
new_train_X.columns = remaining_features
new_test_X.columns = remaining_features
```

```python
new_train_X.head()
# looks like our best ones are the actual features we had
```

```python
log_reg = LogisticRegression()
log_reg = log_reg.fit(new_train_X, train_y)
log_reg.score(new_test_X, test_y)
```

```python
# looks like we improved our accuracy despite losing features
```

```python
# open up the files
opened_red_data = open('/content/winequality-red.csv','r')
opened_white_data = open('/content/winequality-white.csv','r')

# read the data and remove the names
red_data_lines = opened_red_data.readlines()
opened_red_data.close()
red_data_lines.pop(0)

# do the same for the white wines
white_data_lines = opened_white_data.readlines()
opened_white_data.close()
names = white_data_lines.pop(0)
```

```python
# if the name type is not string (it is list) then we don't want to try this
if type(names) == str:
  # remove the " at the beginning and the "\n at the end.
  names = names[1:-2]
  names = names.split('";"')
  # add type to the names
  names.append("type")
names
```

```python
# here we make cleaned_red_data to eventually put all of our data
cleaned_red_data = []

for line in red_data_lines:
  # the data is separated by semicolons, so we split by those
  split_line = line.split(';')

  # for each of the features, we convert it into numbers
  for i in range(12):
    split_line[i] = float(split_line[i])

  # here, we add 0 for the class
  # 0 is for red wine
  split_line.append(0)

  # we add the line to cleaned_red_data
  cleaned_red_data.append(split_line)
```

```python
# we do the same for white data
cleaned_white_data = []

for line in white_data_lines:
  split_line = line.split(';')

  for i in range(12):
    split_line[i] = float(split_line[i])

  # 1 is for white wine
  split_line.append(1)

  cleaned_white_data.append(split_line)
```

```python
# import pandas under the nickname of pd
import pandas as pd

# add the lists together and join them into a dataframe
all_data = cleaned_red_data + cleaned_white_data
data = pd.DataFrame(all_data)
data.head()
```

```python
# shuffle the data, using a random state of 0
data = data.sample(frac=1,random_state=0)
# here, we reset the indices to be in order and we don't want to get
# the old indices as a new column, so drop = True
data = data.reset_index(drop=True)
# give the columns names
data.columns = names
data.head()
```

```python
X = data.iloc[:, 0:12]
X.head()
```

```python
y = data["type"]
```

```python
# separate out train, validation, and test x and y
train_X = X.iloc[:-600]
valid_X = X.iloc[-600:-200]
test_X = X.iloc[-200:]
train_y = y.iloc[:-600]
valid_y = y.iloc[-600:-200]
test_y = y.iloc[-200:]
```

```python
# import relevant libraries
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectKBest
```

```python
# here, we have a function to run a model based on some particular hyperparameters
def run_model(feature_number, tree_count, max_dep):
  # we want to select the feature_number best features
  fil = SelectKBest(k=feature_number)
  fil = fil.fit(train_X, train_y)
  new_train_X = fil.transform(train_X)
  new_valid_X = fil.transform(valid_X)
  # make and train a random forest on training data then return the accuracy
  rf = RandomForestClassifier(n_estimators = tree_count, max_depth = max_dep)
  rf.fit(new_train_X, train_y)
  return rf.score(new_valid_X, valid_y)
```

```python
# here we just define the posibilities for what hyperparameters we want
# you probably have different combinations depending on what you chose
feature_number_pos = [i for i in range(1, 11)]
tree_count_pos = [1, 5, 25, 100]
max_dep_pos = [1, 5, 10, 20]
```

```python
# we want to get a list of the good combos
good_combos = []
# a triple nested for loop goes through all possible combos of hyperparameters
for fn in feature_number_pos:
  for tc in tree_count_pos:
    for md in max_dep_pos:
      # get the accuracy and print it out
      acc = run_model(fn, tc, md)
      # if we have good accuracy add it to our good combos
      if acc>=0.9875:
        print("good combo:")
        print("number of features:", fn)
        print("tree count", tc)
        print("max depth", md)
        print()
        good_combos.append([fn, tc, md])
```

```python
# print out our good combos
print(good_combos)
```

```python
for gc in good_combos:
  fil = SelectKBest(k=gc[0])
  fil = fil.fit(train_X, train_y)
  new_train_X = fil.transform(train_X)
  new_test_X = fil.transform(test_X)
  # make and train a random forest on training data then return the accuracy
  rf = RandomForestClassifier(n_estimators = gc[1], max_depth = gc[2])
  rf.fit(new_train_X, train_y)
  print(gc, rf.score(new_test_X, test_y))
```
