```python
import pandas as pd
```

```python
df = pd.read_csv("/content/patient_dataset.csv", header = None)
df.head()
# first column is the feature: age
# we are trying to use the age to predict whether
# the patient will recover from the disease (second column)
```

```python
num_patients = len(df)
print(num_patients)
```

```python
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression()
```

```python
# this method is called leave-one-out

total_score = 0
# start the total score at 0 and iterate through patients
for i in range(num_patients):
  # we get all the patients before i (0:i) and all the patients after (i+1:)
  # and combine them together as our training dataset
  train = pd.concat( [ df.iloc[:i], df.iloc[i+1:] ] )
  # then we have the testing sample
  test = df.iloc[i]
  # separate into X and y to train the model

  # colon before the comma means we are getting all of the rows
  # the range after the comma is the columns we want
  # for train_X we want all the columns before the last one
  train_X = train.iloc[:,:1]
  # and for train_y we want just the last column
  train_y = train.iloc[:,1]
  log_reg.fit(train_X, train_y)
  # if the predicted output is correct
  if (log_reg.predict( [ test.iloc[:1] ] ) == test[1]):
    total_score += 1
    # add one to the total score

print("Average score:", total_score / num_patients)
```
